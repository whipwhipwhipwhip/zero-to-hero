```python
import math
import numpy as np
import matplotlib.pyplot as plt
import torch
from sklearn.model_selection import train_test_split
%matplotlib inline
```

```python
words = open('../../../src/zero_to_hero/part_02_makemore/names.txt', 'r').read().splitlines()
```

```python
seed=42
train, test = train_test_split(words, test_size=0.1, random_state = seed)
```

```python
## Important, we have started using cross-entropy from here!!
from zero_to_hero.part_02_makemore.makemore import NgramEstimator
from sklearn.model_selection import ShuffleSplit, cross_val_score
search = ShuffleSplit(n_splits=1, test_size=1/9, random_state=42)

min_score = -np.inf
best_lam = 0
for lam in np.logspace(-3, 2, 20):
    print(f"Trialling {lam}")
    score = cross_val_score(NgramEstimator(regularisation=lam, epochs=200), X=train, cv=search)
    if score > min_score:
        best_lam = lam
        min_score = score
        print(f"New lambda of {best_lam}, new score of {min_score}")
```

```python
# Now evaluate the best
test_loss = -NgramEstimator(regularisation=best_lam).fit(X=train).score(X=test)
print(f"test loss {test_loss:.4f}")
```

## Bigram

```python
from zero_to_hero.part_02_makemore.makemore import NgramEstimator
from sklearn.model_selection import ShuffleSplit, cross_val_score
search = ShuffleSplit(n_splits=1, test_size=1/9, random_state=42)

min_score = -np.inf
best_lam = 0
for lam in np.logspace(-3, 2, 20):
    print(f"Trialling {lam}")
    score = cross_val_score(NgramEstimator(regularisation=lam, epochs=200, gram=2), X=train, cv=search)
    if score > min_score:
        best_lam = lam
        min_score = score
        print(f"New lambda of {best_lam}, new score of {min_score}")
```

```python
# Now evaluate the best
test_loss = -NgramEstimator(regularisation=best_lam, gram=2).fit(X=train).score(X=test)
print(f"test loss {test_loss:.4f}")
```
