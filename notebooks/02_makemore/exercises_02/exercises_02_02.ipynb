## Exercise 2 - doing a proper train/test split

```python
import math
import numpy as np
import matplotlib.pyplot as plt
import torch
from sklearn.model_selection import train_test_split
%matplotlib inline
```

```python
words = open('../../../src/zero_to_hero/part_02_makemore/names.txt', 'r').read().splitlines()
```

```python
train, test = train_test_split(words, test_size=0.1)
```

```python
train, train_val = train_test_split(words, test_size=1/9)
```

```python
from zero_to_hero.part_02_makemore.makemore import train_ngram, evaluate_loss
W, ctx_stoi, char_stoi, char_itos = train_ngram(
    train, gram=3, epochs=200, lr=50,
    reg=0.01, seed=2147483647,
)
```

```python
val_loss = evaluate_loss(train_val, 3, W, ctx_stoi, char_stoi)
print(f"val loss {val_loss:.4f}")
```

```python
test_loss = evaluate_loss(test, 3, W, ctx_stoi, char_stoi)
print(f"test loss {test_loss:.4f}")
```
