## Using tiktoken and endoftext special char

```python
%load_ext autoreload
%autoreload 2
```

```python
import tiktoken
enc = tiktoken.get_encoding("cl100k_base") # this is the GPT-4 tokenizer
ids = enc.encode("<|endoftext|>hello world", allowed_special="all")
```

```python
from nanotok import TiktokenBPETokenizer
```

```python
gpt4_split = r"""'(?i:[sdmt]|ll|ve|re)|[^\r\n\p{L}\p{N}]?+\p{L}+|\p{N}{1,3}| ?[^\s\p{L}\p{N}]++[\r\n]*|\s*[\r\n]|\s+(?!\S)|\s+"""
special_tokens = ["<|endoftext|>"]
tok = TiktokenBPETokenizer("cl100k_base", split_pattern=gpt4_split, special_tokens=special_tokens)
```

```python
st = "<|endoftext|>hello world"
my_enc = tok.encode(st)
```
