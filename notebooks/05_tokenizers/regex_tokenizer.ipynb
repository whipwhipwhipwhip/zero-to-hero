```python
regex_groups = regex.findall(GPT4_SPLIT_PATTERN, text)
```

```python
from collections import defaultdict
def find_largest_global(groups):
    d = defaultdict(int)
    top, top_pair = 0, 0
    # can use zip(st, st[1:])
    # just build dict and do max(d, key=d.get)
    for group in groups:
        for i in range(len(group)-1):
            pair = (group[i], group[i+1])
            d[pair] += 1
            if d[pair] > top:
                top, top_pair = d[pair], pair

    return top, top_pair
```
